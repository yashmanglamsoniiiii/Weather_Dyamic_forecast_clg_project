# Weather Forecast ML - Exploratory Data Analysis (EDA)
## Dataset: ECMWF Copernicus ERA5 Hourly Reanalysis
This notebook demonstrates exploratory data analysis, atmospheric feature relationships, class imbalance handling, and diurnal/seasonal patterns.

```python
import pandas as pd
import numpy as np
import os

DATA_PATH = '../data/raw/era5_hourly_data.csv'
df = pd.read_csv(DATA_PATH)
df['time'] = pd.to_datetime(df['time'])
print(f'Total observations: {len(df):,}')
print(f'Monitored cities: {df["city"].unique()}')
df.head()
```

### 1. Meteorological Variables & Descriptive Statistics
Variables include temperature, relative humidity, dew point, surface pressure, cloud cover, and wind speed.

```python
stats = df[['temperature_2m', 'relative_humidity_2m', 'dew_point_2m', 'surface_pressure', 'wind_speed_10m', 'rain']].describe()
print(stats)
```

### 2. Rain Event Imbalance Analysis
Precipitation occurs in fewer hours compared to dry atmospheric states. We quantify this imbalance to guide scale_pos_weight tuning.

```python
rain_count = (df['rain'] > 0.05).sum()
dry_count = len(df) - rain_count
print(f'Dry hours: {dry_count:,} ({dry_count / len(df) * 100:.1f}%)')
print(f'Rain hours: {rain_count:,} ({rain_count / len(df) * 100:.1f}%)')
print(f'Imbalance Ratio: {dry_count / rain_count:.2f}x (addressed via scale_pos_weight in XGBoost)')
```

### 3. Atmospheric Feature Correlation Matrix
High relative humidity and lower dew point depression strongly correlate with precipitation events.

```python
cols = ['temperature_2m', 'relative_humidity_2m', 'dew_point_2m', 'surface_pressure', 'cloud_cover', 'wind_speed_10m', 'rain']
corr = df[cols].corr()
print(corr.round(3))
```
